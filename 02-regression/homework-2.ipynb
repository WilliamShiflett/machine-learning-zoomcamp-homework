Import some data and read it into a dataframe.

```python
import pandas as pd
import numpy as np
```

```python
data = 'https://raw.githubusercontent.com/alexeygrigorev/datasets/master/car_fuel_efficiency.csv'
```

```python
!wget $data
```

```python
df = pd.read_csv('car_fuel_efficiency.csv')
```

```python
df = df[['engine_displacement','horsepower','vehicle_weight','model_year','fuel_efficiency_mpg']]
```

Check the distribution of the fuel_efficiency_mpg variable.

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Create a figure and a grid of subplots
fig, axes = plt.subplots(2, 3, figsize=(10, 8))

# Plot histograms on each subplot
sns.histplot(data=df, x="fuel_efficiency_mpg", ax=axes[0, 0])
sns.histplot(data=df, x="model_year", ax=axes[1, 0])
sns.histplot(data=df, x="vehicle_weight", ax=axes[0, 1])
sns.histplot(data=df, x="horsepower", ax=axes[1, 1])
sns.histplot(data=df, x="engine_displacement", ax=axes[0, 2])
```
